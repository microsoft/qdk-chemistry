# Explore the active orbitals of stretched N₂

This notebook complements the active-space chapter with interactive views of the atom-centered basis functions and the candidate natural orbitals considered by autoCAS. You will first relate each basis function to its nucleus and angular shape, then compare each natural orbital's three-dimensional shape with its natural occupation, single-orbital entropy, and final selection status.

The numerical workflow is defined in `tutorial_choose_active_space.py`, which must remain in the same folder as this notebook. Sharing that implementation ensures that the command-line example, interactive visualization, and automated tests all perform the same chemistry calculation.

The import cell silences QDK/Chemistry library logs so the numerical results remain easy to read. Change `Logger.LogLevel.off` to `Logger.LogLevel.info` and rerun the cell to see detailed calculation logs.

```python
from qdk.widgets import MoleculeViewer
from qdk_chemistry.utils import Logger
from tutorial_choose_active_space import (
    create_stretched_n2_structure,
    generate_active_orbital_cube_data,
    generate_basis_function_cube_data,
    plot_orbital_entropy_selection,
    print_active_space_results,
    run_active_space_workflow,
)

Logger.set_global_level(Logger.LogLevel.off)
```

## Inspect the basis functions

A basis set supplies localized spatial functions centered on the nuclei. Molecular orbitals are linear combinations of these basis functions.

The next viewer shows each cc-pVDZ basis function separately. Use the menu labels and metadata to identify its nitrogen center and angular type, then compare its angular shape and radial extent. For $p$ and $d$ functions, a node passes through the nucleus, so the visible lobes and peak amplitude lie away from it even though the function is centered there.

Generating the basis-function grids may take a moment.

```python
structure = create_stretched_n2_structure()
basis_function_cube_data = generate_basis_function_cube_data(
    structure,
    "cc-pvdz",
    grid_size=(30, 30, 30),
    margin=3.0,
)

MoleculeViewer(
    molecule_data=structure.to_xyz(),
    cube_data=basis_function_cube_data,
)
```

## Select the active space

Run the complete stretched-N₂ workflow from the chapter. The calculation starts from the cc-pVDZ Hartree–Fock wavefunction, solves the initial valence-space problem, transforms the candidate orbitals to natural orbitals, uses their single-orbital entropies to select the refined active space, and solves the refined active-space Hamiltonian.

The printed results provide the energies, determinant counts, orbital entropies, and final orbital partition to record in your lab notebook.

```python
result = run_active_space_workflow()
print_active_space_results(result)
```

## Compare the orbital entropies

The next plot sorts the candidate natural orbitals by decreasing single-orbital entropy. The dashed line separates the high-entropy group retained by autoCAS from the excluded orbitals. The x-axis labels preserve the original natural-orbital indices.

The molecular viewer needs values of each orbital's spatial wavefunction on a three-dimensional grid. The next cell evaluates all candidate natural orbitals on such a grid and stores the sampled values as cube data.

Each orbital is also annotated with its natural occupation, single-orbital entropy, and autoCAS selection status so that you can examine the spatial and numerical evidence together. Generating the orbital grids may take a little longer than the preceding calculation.

```python
cube_data = generate_active_orbital_cube_data(
    result,
    grid_size=(30, 30, 30),
    margin=10.0,
)
```
